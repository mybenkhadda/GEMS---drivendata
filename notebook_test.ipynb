```python

import tifffile
import numpy as np
import matplotlib.pyplot as plt

# Read the TIFF file
image = tifffile.imread("gems-geodawn-numerical-features.tif")

print("Image shape:", image.shape)

# Assume image shape is (height, width, channels)
# Convert channel-first images (channels, height, width)
if image.ndim == 3 and image.shape[0] <= 4 and image.shape[-1] > 4:
    image = np.moveaxis(image, 0, -1)

# Plot channels
if image.ndim == 2:
    plt.imshow(image, cmap="gray")
    plt.title("Grayscale image")
    plt.axis("off")
else:
    n_channels = image.shape[-1]
    fig, axes = plt.subplots(
        1, n_channels, figsize=(5 * n_channels, 5),
        squeeze=False
    )

    for i in range(n_channels):
        ax = axes[0, i]
        ax.imshow(image[:, :, i], cmap="gray")
        ax.set_title(f"Channel {i + 1}")
        ax.axis("off")

    plt.tight_layout()

plt.show()
```
